Welcome to my daily Rosalind grind. This repository is a living record of my solutions to problems on [Rosalind](https://rosalind.info/), a platform for learning bioinformatics through problem-solving.

I'm a BS-MS student at IISER Bhopal interested broadly in compuational biology and bioinformatics. 

**What to expect here:**
- Solutions in Python (`.ipynb` files, day-wise)
- Clear explanations of the biological and mathematical reasoning
- Occasional inefficient implementations (I'll flag these and revisit them as I learn better approaches)
- A genuine learning log, not a polished showcase

> *"Correct before clever."* Some of my early solutions will brute-force things that could be done elegantly. That's okay, I'm documenting the journey, not the destination.

To keep bioinformatics accessible to everyone, Rosalind problems intentionally require zero biological prerequisites. Consequently, they often abstract away the messy variables and intricacies of real-world biological systems. One of my core goals for this repository is to revisit these foundational problems as my programming acumen grows, reintroducing those missing biological complexities to design, simulate, and solve more biologically realistic challenges.

Lastly, as someone new to programming, I often need to dive into the theory behind crucial algorithms and concepts, like time complexity (to write more efficient code), dynamic programming, etc. To keep track of what I learn, I’ll be updating this space with high-quality resources and my personal study notes. I hope you find them helpful too! Click [here](https://drive.google.com/drive/folders/1MnFKbVpq3meN7FQ-Esdkc0NpqFCk-dFB?usp=drive_link) for the notes!
