Our aim is to somehow modify this recurrence relation to achieve a dynamic programming solution in the case that all rabbits die out after a fixed number of months. Given: Positive integers n≤100
 and m≤20

Return: The total number of pairs of rabbits that will remain after the n
-th month if all rabbits live for m
 months.

# Revisiting the rabbit problem from Day01

def fib_rabbit(n,k): <br>
    if n==1 or n==2: # only rabbits alive two months before can reproduce <br>
        return 1

    return (k*(fib_rabbit(n-2,k))+fib_rabbit(n-1,k)) # each reproducing rabbit pair produces k rabbit pairs

```python
# i tried solving this using the tabular method for solving DP problems

def mortal_fib_rabbit(n,m):
    dp=[0]*(n+1)
    dp[0]=1
    dp[1]=1
    dp[2]=1

    for i in range(3,m+1):
        dp[i]=dp[i-2]+dp[i-1]
    for x in range(m+1,n+1):
        dp[x]=dp[x-2]+dp[x-1]-dp[x-m-1]
    return dp[n]
    
    

mortal_fib_rabbit(87,19)
   
    
   

```

###  Mortal Fibonacci Rabbits (Dynamic Programming)

### 1. Phase 1: Growth Period ($i \le m$)
During the first $m$ months, no rabbits have died yet because the oldest generation hasn't reached its lifespan limit.
* **Logic:** The population grows exactly like the standard Fibonacci sequence.
* **Formula:** `dp[i] = dp[i - 1] + dp[i - 2]`

### 3. Phase 2: Mortality Period ($i > m$)
Once the timeline passes $m$ months, rabbits begin to die off.
* **Logic:** We take the total expected pairs (survivors + newborns) and subtract the generation that dies this month.
* **The Death Shift:** The rabbits dying in month $i$ are the ones born exactly $m$ months ago. To remove them accurately, we subtract `dp[i - m - 1]`.
* **Formula:** `dp[x] = dp[x - 2] + dp[x - 1] - dp[x - m - 1]`



```python
# just for practice, the day01 problem using dp

def normal(n,k):
    dp=[0]*(n+1)
    dp[1]=1
    dp[2]=1

    for i in range(3,n+1):
        dp[i]=k*dp[i-2]+dp[i-1]
    return dp[n]


normal(5,3)
    
```

```python
# fibonacci using dp

def fib(n):
    dp=[0]*(n+1)
    dp[0]=1
    dp[1]=1

    for i in range(2,n+1):
        dp[i]=dp[i-1]+dp[i-2]
    return dp

fib(5)
```
