# Problem Statement: Consensus and Profile Matrix

A **matrix** is a rectangular table of values divided into rows and columns.  
An \(m \times n\) matrix has \(m\) rows and \(n\) columns.

Given a matrix \(A\), we write \(A_{i,j}\) to indicate the value found at the intersection of row \(i\) and column \(j\).

---

## Profile Matrix

Suppose we have a collection of DNA strings, all having the same length \(n\).

Their **profile matrix** is a \(4 \times n\) matrix \(P\), where:

- \(P_{1,j}\) represents the number of times that `A` occurs in the \(j\)-th position
- \(P_{2,j}\) represents the number of times that `C` occurs in the \(j\)-th position
- \(P_{3,j}\) represents the number of times that `G` occurs in the \(j\)-th position
- \(P_{4,j}\) represents the number of times that `T` occurs in the \(j\)-th position

---

## Consensus String

A **consensus string** is a string of length \(n\) formed by taking the most common symbol at each position.

The \(j\)-th symbol of the consensus string corresponds to the symbol having the maximum value in the \(j\)-th column of the profile matrix.

If several symbols are tied for the most common value, then any one of them may be chosen.

---

## Example

Given the following DNA strings:

```text
ATCCAGCT
GGGCAACT
ATGGATCT
AAGCAACC
TTGGAACT
ATGCCATT
ATGGCACT
```

The profile matrix is:

```text
A: 5 1 0 0 5 5 0 0
C: 0 0 1 4 2 0 6 1
G: 1 1 6 3 0 1 0 0
T: 1 5 0 0 0 1 1 6
```

The consensus string is:

```text
ATGCAACT
```

---

## Problem

### Given

A collection of at most 10 DNA strings of equal length, at most 1 kbp, in FASTA format.

### Return

A consensus string and profile matrix for the collection.
